## **1. Histogram**

**Understand the Concept**

**Purpose**

* Histogram shows how numerical data is distributed.
* It groups values into ranges and shows how often they occur.

**When to Use**

* Use a histogram when we want to understand the distribution of numerical data.
* It can help us see the center, spread, and unusual values.

**Advantages**

* Easy to understand.
* Shows the shape of the data.
* Helps identify skewness and possible outliers.

**Limitations**

* It is mainly used for numerical data.
* The graph can look different depending on the number of bins.



**Demonstrate the Concept**

**Real-Life Example:**

* A gym records the weight of its members.
* A histogram can show how many members fall into different weight ranges.



```python
# Implement the Concept:
import matplotlib.pyplot as plt

# Weight data
weights = [55, 58, 60, 70, 72, 74, 75, 78, 80, 82]

# Histogram
plt.hist(weights, bins=10,  edgecolor='k')
plt.xlabel("Weight (kg)")
plt.ylabel("Number of Members")
plt.title("Distribution of Gym Members' Weight")
plt.show()
```

**Interpretation**

* The histogram shows how the members' weights are distributed.
* Each bar represents a weight range.
* Taller bars mean more members fall in that weight range.

## **2. Bar Chart**

**Understand the Concept**
**Purpose**

* Bar chart is used to compare different categories.
* The height of each bar shows the value of that category.

**When to Use**

* Use a bar chart when we want to compare categories.

**Advantages**

* Easy to compare categories.
* Simple to read.
* Works well with small and large groups of categories.

**Limitations**

* Not suitable for showing continuous data.
* Too many categories can make the chart difficult to read.





**Demonstrate the Concept**

**Real-Life Example:**

* A shop records the number of products sold in one day.
* A bar chart can compare the sales of different products.



```python
# Implement the Concept:
# Product sales
products = ["Rice", "Milk", "Bread", "Eggs"]
sales = [50, 35, 45, 60]

# Bar chart
bar = plt.bar(products, sales)
plt.bar_label(bar) # label for chart
plt.xlabel("Products")
plt.ylabel("Number of Sales")
plt.title("Daily Product Sales")
plt.show()
```

**Interpretation**

* Eggs have the highest sales with 60.
* Milk has the lowest sales with 35.
* The bar chart makes it easy to compare the sales of each product.

## **3. Line Chart**



**Understand the Concept**

* Line chart shows how a value changes over time or in a sequence.
* The points are connected with lines to show the trend.

**Purpose**

* Used to show changes and trends over time.
* Helps us see whether values are increasing or decreasing.

**When to Use**

* Use a line chart for time-based data.

**Advantages**

* Easy to see trends.
* Shows increases and decreases clearly.
* Useful for comparing changes over time.

**Limitations**

* Not suitable for unrelated categories.
* Too many lines can make the chart difficult to understand.



**Demonstrate the Concept**

**Real-Life Example:**

* A person records their daily water intake for one week.
* A line chart can show how the water intake changes each day.


```python
# Implement the Concept:
# water intake data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
water_intake = [2.0, 2.5, 2.2, 2.8, 2.4, 3.0, 2.7]

# line chart
plt.plot(days, water_intake, marker="o")

plt.xlabel("Day")
plt.ylabel("Water Intake (Liters)")
plt.title("Daily Water Intake")
plt.show()
```

**Interpretation**

* Water intake changes from day to day.
* Saturday has the highest water intake at 3 liters.
* Monday has the lowest at 2 liters.

## **4. Scatter Plot**



**Understand the Concept**

* Scatter plot shows the relationship between two numerical variables.
* Each point represents one observation.
* It helps us see patterns, relationships, and unusual values.

**Purpose**

* Used to show the relationship between two variables.
* Helps us see whether the variables move together or in opposite directions.

**When to Use**

* Use a scatter plot when working with two numerical variables.
* Examples include study hours and marks, height and weight, or advertising and sales.

**Advantages**

* Easy to see relationships and patterns.
* Helps identify possible outliers.
* Useful for checking positive or negative relationships.

**Limitations**

* Not suitable for categorical data.
* Too many points can make the graph difficult to read.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher records students' study hours and exam marks.
* A scatter plot can show whether students who study more usually get higher marks.


```python
# Implement the Concept:
# Study hours and exam marks data
study_hours = [2, 4, 6, 8, 9, 10]
exam_marks = [50, 60, 70, 80, 55, 90]

# scatter plot
plt.scatter(study_hours, exam_marks)

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.title("Study Hours vs Exam Marks")
plt.show()
```

**Interpretation**

* Most points show that marks increase when study hours increase.
* The point at 9 hours and 55 marks is far from the general pattern.
* This point can be considered a possible outlier.
* The scatter plot helps us see both the relationship and unusual points.

## **5. Pie Chart**



**Understand the Concept**

* Pie chart shows how a total is divided into different categories.
* Each slice represents one category.
* Bigger slices mean a bigger share of the total.

**Purpose**

* Used to show percentages or parts of a whole.
* Helps us understand which category has a larger or smaller share.

**When to Use**

* Use a pie chart when there are only a few categories.
* It is useful for data like expenses, food choices, or product sales.

**Advantages**

* Easy to understand the share of each category.
* Good for showing simple percentage comparisons.
* Gives a quick view of the whole data.

**Limitations**

* Not suitable for many categories.
* Similar-sized slices can be difficult to compare.
* It is not good for showing changes over time.



**Demonstrate the Concept**

**Real-Life Example:**

* A student tracks how they spend their monthly pocket money.
* The money is divided into Food, Travel, Shopping, and Savings.
* A pie chart can show the share of each expense.



```python
# Implement the Concept:
# pocket money
categories = ["Food", "Travel", "Shopping", "Savings"]
amounts = [3000, 1500, 1000, 2500]

# pie chart
plt.pie(amounts, labels=categories, autopct="%1.1f%%")

plt.title("Monthly Pocket Money Distribution")
plt.show()
```

**Interpretation**

* Food takes the largest share of the monthly money.
* Shopping takes the smallest share.

## **6. Density Plot**



**Understand the Concept**

* Density plot shows how numerical data is distributed.
* It shows where the data values are more concentrated.
* It gives a smooth view of the data distribution.

**Purpose**

* Used to understand the shape and spread of numerical data.
* Helps identify where most of the values are concentrated.

**When to Use**

* Use a density plot for numerical data.
* It is useful when comparing the distribution of two or more groups.

**Advantages**

* Gives a smooth view of data distribution.
* Easy to compare different groups.
* Helps identify the center and shape of the data.

**Limitations**

* Not suitable for categorical data.
* The shape can change based on the smoothing method.
* It may be harder to understand for beginners compared with a histogram.



**Demonstrate the Concept**

**Real-Life Example:**

* A gym records the weight of its members.
* A density plot can show where most members weights are concentrated.



```python
# Implement the Concept:
import seaborn as sns

# Weight 
weights = [55, 58, 60, 62, 64, 65, 67, 68, 70, 72, 74, 75, 78, 80, 82]

# Create density plot
sns.kdeplot(weights, fill=True)
plt.xlabel("Weight (kg)")
plt.ylabel("Density")
plt.title("Density Plot of Gym Members' Weight")
plt.show()
```

**Interpretation**

* The highest part of the curve shows where most members' weights are concentrated.
* The curve gives a smooth view of the weight distribution.
* It helps us understand the shape of the data more easily.

## **7. Box Plot**



**Understand the Concept**

* Box plot shows how numerical data is spread.
* It shows the median, lower values, higher values, and possible outliers.
* It is useful for quickly understanding the distribution of data.

**Purpose**

* Used to show the spread and center of numerical data.
* Helps identify possible outliers.

**When to Use**

* Use a box plot when you want to compare the distribution of different groups.
* It is useful for salary, marks, age, weight, and other numerical data.

**Advantages**

* Easy to identify outliers.
* Shows the spread of data clearly.
* Useful for comparing multiple groups.

**Limitations**

* Does not show every individual value.
* It may be difficult for beginners to understand all parts of the plot.



**Demonstrate the Concept**

**Real-Life Example:**

* A company records the salaries of its employees.
* A box plot can show the middle salary range and identify unusually high or low salaries.

**AI/ML Example**

* Box plots are useful during EDA.
* They help find the spread of features and possible outliers before training a Machine Learning model.


```python
# Implement the Concept:
# Employee salaries
salaries = [20000, 25000, 28000, 30000, 32000, 35000, 40000, 45000, 50000, 150000]

# box plot
plt.boxplot(salaries)

plt.ylabel("Salary (₹)")
plt.title("Employee Salary Distribution")
plt.show()
```

**Interpretation**

* The box shows the middle part of the salary data.
* The line inside the box shows the median.
* The value ₹1,50,000 is far from the other salaries.
* The box plot helps us identify this possible outlier.

## **8. Violin Plot**

**Understand the Concept**

* Violin plot shows the distribution and spread of numerical data.
* It combines ideas from a box plot and a density plot.
* The wider part shows where more data values are concentrated.

**Purpose**

* Used to understand the shape and spread of numerical data.
* Helps compare the distribution of different groups.

**When to Use**

* Use a violin plot when we want to compare numerical data between groups.
* It is useful for marks, salaries, weights, or test scores.

**Advantages**

* Shows the spread and shape of the data.
* Shows where values are more concentrated.
* Useful for comparing multiple groups.

**Limitations**

* More difficult to understand than a simple bar chart.
* Not suitable when there are very few data points.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher wants to compare the marks of students from two classes.
* A violin plot can show how the marks are distributed in each class.



```python
# Implement the Concept:
# Marks of two classes
class_a = [55, 60, 65, 70, 72, 75, 78, 80, 82]
class_b = [60, 62, 65, 68, 70, 72, 74, 76, 78]

# Create data
data = {"Class": ["Class A"] * len(class_a) + ["Class B"] * len(class_b),
        "Marks": class_a + class_b}

#  violin plot
sns.violinplot(x="Class", y="Marks", data=data)
plt.title("Marks Distribution of Two Classes")
plt.show()
```

**Interpretation**

* The violin plot shows how the marks are distributed in both classes.
* Wider areas show where more students have marks.
* It helps compare the spread and shape of marks between Class A and Class B.

## **9. Pair Plot**



**Understand the Concept**

* Pair plot shows the relationship between multiple numerical variables.
* It creates small graphs for each pair of variables.
* It also shows the distribution of each variable.

**Purpose**

* Used to study relationships between several variables at the same time.
* Helps us find patterns, correlations, and possible outliers.

**When to Use**

* Use a pair plot when a dataset has multiple numerical features.
* It is useful during Exploratory Data Analysis.

**Advantages**

* Shows many relationships in one graph.
* Helps find positive, negative, or no relationships.
* Useful for finding possible outliers.

**Limitations**

* Can become difficult to read when there are many features.
* It can take more time to generate for large datasets.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher records students' study hours, attendance, and exam marks.
* A pair plot can show the relationship between each pair of these values.

**AI/ML Example**

* Pair plots are commonly used during EDA.
* They help us understand relationships between features before training a ML model.



```python
# Implement the Concept:
import pandas as pd
# Student data
data = pd.DataFrame({"Study_Hours": [2, 4, 6, 8, 10],
                     "Attendance": [60, 70, 75, 85, 95],
                     "Exam_Marks": [50, 60, 70, 80, 90]})

# Pair plot
sns.pairplot(data)
plt.show()
```

## **10. Heatmap**



**Understand the Concept**

* Heatmap shows data using different shades or colors.
* It makes high and low values easy to identify.
* A correlation heatmap is commonly used to show relationships between variables.

**Purpose**

* Used to display values in an easy visual form.
* Helps quickly identify strong and weak relationships between variables.

**When to Use**

* Use a heatmap when working with a correlation matrix or other table-like numerical data.
* It is useful when we want to compare many values at the same time.

**Advantages**

* Easy to identify patterns.
* Shows many values in one graph.
* Useful for finding strong correlations quickly.

**Limitations**

* Too many variables can make the heatmap difficult to read.
* The colors must be chosen carefully so the values are easy to understand.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher records the study hours, attendance, and exam marks of students.
* A heatmap can show how strongly these variables are related.



```python
# Implement the Concept:
# Student data
data = pd.DataFrame({"Study_Hours": [2, 4, 6, 8, 10],
                     "Attendance": [60, 70, 75, 85, 95],
                     "Exam_Marks": [50, 60, 70, 80, 90]})

# correlation matrix
correlation_matrix = data.corr()

# heatmap
sns.heatmap(correlation_matrix, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

**Interpretation**

* The heatmap shows the correlation between study hours, attendance, and exam marks.
* Values close to +1 show a strong positive relationship.
* Values close to 0 show little or no linear relationship.
* The colors make it easier to identify strong relationships quickly.
