# **Notebook 8 – Correlation & Covariance**

## **1. Correlation**



**Understand the Concept**

* Correlation shows the relationship between two variables.
* It tells us whether the variables move together or in opposite directions.
* The correlation value is between -1 to +1.



**Demonstrate the Concept**

**Real-Life Example:**

* A student studies more hours and usually gets higher marks.
* So, study hours and marks may have a positive relationship.
  
**Example:**

* Study hours: 2, 4, 6, 8, 10
* Exam marks: 50, 60, 70, 80, 90
* When study hours increase, exam marks also increase.
* So, these two variables have a positive correlation.


```python
# Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Implement the Concept:

# Data - study hours & exam marks
data = pd.DataFrame({"Study_Hours": [2, 4, 6, 8, 10],
                     "Exam_Marks": [50, 60, 70, 80, 90]})

# Correlation
print(data.corr())
```

## **2. Covariance**



**Understand the Concept**

* Covariance shows how two variables change together.
* Positive covariance - both variables usually increase or decrease together.
* Negative covariance - one variable increases while the other decreases.

**Formula :** Covariance = Σ[(X - X̄)(Y - Ȳ)] / (n - 1)



**Demonstrate the Concept**

**Real-Life Example:**

* A shop records the number of advertisements and daily sales.
* If more advertisements are followed by higher sales, the covariance will be positive.


```python
# Implement the Concept:
# Data - ads & sales
data_1 = pd.DataFrame({"ads" : [2,4,6,8],
                     "sales" : [3000,5000,7000,9000]})
# Covariance
covariance = data_1["ads"].cov(data_1["sales"])
print(covariance)
```

It is positive, so ads count and sales amount move in the same direction.

## **3. Positive Correlation**



**Understand the Concept**

* Positive correlation means two variables move in the same direction.
* When one variable increases, the other variable also increase.
* When one variable decreases, the other may also decrease.
* The correlation value is between 0 and +1.



**Demonstrate the Concept**

**Real-Life Example:**

* A shop spends more money on advertising and gets more customers.
* So, advertising and number of customers can have a positive correlation.



```python
# Implement the Concept:
# Data - ads cost and no of customers
data_2 = pd.DataFrame({"Advertising_Cost": [10, 20, 30, 40, 50],
                     "Customers": [100, 120, 140, 160, 170]})

# Correlation
correlation = data_2["Advertising_Cost"].corr(data_2["Customers"])

print("Correlation:", correlation)
```

**Explaination:**

* The correlation is 0.99 and ads cost increases, the no of customers also increases.
* So, its a positive correlation.

## **4. Negative Correlation**



**Understand the Concept**

* Negative correlation means two variables move in opposite directions.
* When one variable increases, the other variable is decrease.
* When one variable decrease, the other variable is increases.
* The correlation value is between -1 and 0.



**Demonstrate the Concept**

**Real-Life Example:**

* As the price of a product increases, the number of customers buying it may decrease.
* So, product price and number of customers can have a negative correlation.



```python
# Implement the Concept:
# Data - product price & no of customers
data_3 = pd.DataFrame({"Price" : [300, 350, 400, 450, 500],
                     "Customers": [25,20,15,10,5]}) 
# Negative Correlation
print(data_3.corr())
```

the product price increases, the number of customers decreases.

## **5. No Correlation**



**Understand the Concept**

* No correlation means there is no relationship between two variables.
* The correlation value is close to 0.



**Demonstrate the Concept**

**Real-Life Example:**

* A person's shoe size and their Study Hours usually have no relationship.
* Knowing one does not help us predict the other.
  
**Example:**

* Study Hours: 2, 4, 6, 8, 10
* Shoe Size: 7, 9, 8, 10, 7
* The shoe size does not follow any clear pattern as study hours increase.
* So, there is no clear correlation between these variables.


```python
# Implement the Concept:
# Data - study hours and shoe size
data_4 = pd.DataFrame({ "Study_Hours": [2, 4, 6, 8, 10],
                      "Shoe_Size": [7, 9, 8, 10, 7]})

# No Correlation
correlation = data_4["Study_Hours"].corr(data_4["Shoe_Size"])

print("Correlation:", correlation)
```

There is no clear linear relationship between study hours and shoe size.

## **6. Pearson Correlation**

**Understand the Concept**

* Pearson correlation measures the strength and direction of a linear relationship between two numerical variables.
* Its value is between -1 and +1.
* +1 means perfect positive correlation.
* -1 means perfect negative correlation.
* 0 means no linear correlation.

**Formula :** r = Cov(X, Y) / (σx × σy)



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher checks whether study hours are related to exam marks.
* Pearson correlation can show how strongly the two values are related.
  

```python
# Implement the Concept:
data
```

```python
# Pearson correlation
pearson_correlation = data["Study_Hours"].corr(data["Exam_Marks"], method="pearson")

print("Pearson Correlation:", pearson_correlation)
```

This shows a very strong positive linear relationship.

## **7. Spearman Correlation**



**Understand the Concept**

* Spearman correlation measures the relationship between two variables using their ranks.
* It is useful when the relationship is not a straight line but the values still move in the same direction.
* The value is between -1 and +1.

**Formula :** ρ = 1 - [6Σd² / n(n² - 1)]
* ρ = Spearman correlation
* d = difference between the ranks
* n = number of values



**Demonstrate the Concept**

**AI/ML Example:** Spearman correlation is useful for checking relationships between features when the data is not normally distributed or the relationship is not strictly linear.



```python
# Implement the Concept:
# Spearman correlation
spearman = data["Study_Hours"].corr(data["Exam_Marks"], method="spearman")

print(spearman)
```


A value close to +1 means that students with higher study-hour ranks also have higher mark ranks.

## **8. Correlation Matrix**



**Understand the Concept**

* A correlation matrix is a table that shows the correlation between multiple numerical variables.
* It helps us see how each variable is related to the other variables.
* The values range from -1 to +1.
* Positive values show positive correlation and negative values show negative correlation.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher wants to understand the relationship between study hours, attendance, and exam marks.
  
**Example:**

* Consider three variables: Study Hours, Exam Marks, and Attendance.
* We can calculate the correlation between each pair of variables.


```python
# Implement the Concept:
# Student data
data_5 = pd.DataFrame({"Study_Hours": [2, 4, 6, 8, 10],
                       "Exam_Marks": [50, 60, 70, 80, 90],
                       "Attendance": [60, 70, 75, 85, 95]})

# Correlation matrix
correlation_matrix = data_5.corr()

print(correlation_matrix)
```

## **9. Heatmap**

**Understand the Concept**

* A heatmap is a visual way to show data using different shades or colors.
* In a correlation heatmap, it shows the strength and direction of relationships between variables.
* It makes strong positive and negative correlations easy to identify.



**Demonstrate the Concept**

**Real-Life Example:**

* A teacher wants to quickly understand which student factors are related to exam marks.
* A heatmap can show the relationships between study hours, attendance, and marks.

**AI/ML Example:**

* Heatmaps are commonly used during EDA.
* They help identify highly correlated features before building a ML model.


```python
# Implement the Concept:
# Correlation matrix
correlation_matrix = data_5.corr()
print(correlation_matrix)
```

```python
# Create heatmap
plt.figure(figsize=(8,4))
sns.heatmap(correlation_matrix, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
