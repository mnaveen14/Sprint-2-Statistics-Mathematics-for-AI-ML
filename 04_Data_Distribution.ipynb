## **1. Normal Distribution**

**Understand the Concept**

* Normal distribution is a common way data is spread around the average.
* Most values are close to the mean.
* It has a bell-shaped curve & left and right sides are almost equal.

**Demonstrate the Concept**

**Numerical Example:**

* Suppose the average height of students is 170 cm.
* Most students may have heights close to 170 cm.
* Very short and very tall students are fewer.

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Implement the Concept:
# Height data
heights = np.random.normal(loc=170, scale=8, size=1000)

sns.histplot(data = heights, bins=20,kde=True, color="red") # Use histogram
plt.xlabel("Height")
plt.ylabel("No of Students")
plt.title("Normal Distribution of Student Heights")
plt.show()
```

Most student heights are close to 170 cm.

## **2. Uniform Distribution**

**Understand the Concept**

* Uniform distribution means all values have the same chance of occurring.
* The values are spread evenly between a minimum and maximum value.



**Demonstrate the Concept**

**Real-Life Example:**

* A lucky draw has 10 tickets numbered from 1 to 10.
* Each ticket has the same chance of being selected.
* This is an example of uniform distribution.

```python
# Create tickets data
tickets = np.random.uniform(1,10,100)
print(tickets)
```

```python
sns.histplot(data = tickets, bins = 10, kde = True)
plt.xlabel("data")
plt.ylabel("Frequency")
plt.title("Unifrom Distribution")
plt.show()
```

## **4. Left Skew**



**Understand the Concept**

* Left skew means the data has a longer tail on the left side.
* Most values are toward the higher side.
* Mean < Median < Mode




**Demonstrate the Concept**

**Real-Life Example:**

* A school has an easy exam where most students score high.
* A few students get very low marks.
* This can create a left-skewed distribution.



```python
# Implement the Concept:
marks = np.random.beta(95, 20, 100)

sns.histplot(data = marks, bins = 20, kde = True)
```

* Most of the values are on the higher side.
* The low value creates a tail toward the left.

## **5. Right Skew**



**Understand the Concept**

* Right skew means the data has a longer tail on the right side.
* Most values are toward the lower side.
* Mean > Median > Mode





**Demonstrate the Concept**

**Real-Life Example:**

* A company checks the salaries of its employees.
* Most employees have a normal salary, but a few employees have very high salaries.
* This can create a right-skewed distribution.

```python
# Implement the Concept:
salaries = np.random.beta(18, 120, 100) # 18K
sns.histplot(data = salaries, bins=30, kde = True)
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Right Skewed Distribution")
plt.show()
```

## **7. Kurtosis**

**Understand the Concept**

* Kurtosis tells us about the tails of a distribution.
* It helps us understand whether the data has more or fewer extreme values.

**Key Characteristics**

* **Mesokurtic :** Similar to a normal distribution.
  * Kurtosis = 3
  * Normal level of extreme values.

* **Leptokurtic :** Heavy-tailed distribution.
  * Kurtosis > 3
  * More extreme values than a normal distribution.


* **Platykurtic :** Light-tailed distribution.
  * Kurtosis < 3
  * Fewer extreme values than a normal distribution.


**Demonstrate the Concept**

**Real-Life Example:**

* Stock market returns can have sudden large increases or decreases.
* These extreme changes can create heavy tails in the data.

```python
from scipy.stats import kurtosis
# Data
data = np.array([-5, -2, -1, 0, 1, 2, 3, 15])
# Find kurtosis
kurtosis_value = kurtosis(data, fisher=False)
print(kurtosis_value)
```

```python
# Check type
if kurtosis_value > 3:
    print("Leptokurtic - Heavy-Tailed")
elif kurtosis_value < 3:
    print("Platykurtic - Light-Tailed")
else:
    print("Mesokurtic - Normal")
```

**Or use pandas**

Leptokurtic > 0, Platykurtic < 0 , Mesokurtic = 0

```python
import pandas as pd
data_pd = pd.Series([-5, -2, -1, 0, 1, 2, 3, 15])
print(data_pd.kurt())
```

## **8. Empirical Rule (68–95–99.7)**



**Understand the Concept**

* The Empirical Rule is used for data that follows a normal distribution.
* It tells us how much data is found around the mean.
* About 68% of data falls within 1 standard deviation.
* About 95% of data falls within 2 standard deviations.
* About 99.7% of data falls within 3 standard deviations.



**Demonstrate the Concept**

**Numerical Example:**

* Average exam mark = 70
* Standard deviation = 10

* About 68% of students score between 60 and 80.
* About 95% of students score between 50 and 90.
* About 99.7% of students score between 40 and 100.


```python
mean = 70
std = 10

print("Mean:", mean)
print("1 SD:", mean - std, "to", mean + std)
print("2 SD:", mean - 2*std, "to", mean + 2*std)
print("3 SD:", mean - 3*std, "to", mean + 3*std)
```

Most observations in a normal distribution fall close to the mean, while values far from the mean are uncommon.
