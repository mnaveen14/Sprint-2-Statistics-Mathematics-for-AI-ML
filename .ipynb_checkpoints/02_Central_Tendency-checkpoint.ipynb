## **1. Mean**

**Understand the Concept**

* Mean is the average value of a group of numbers.
* We get the mean by adding all values and dividing the total by the number of values.

**Formula**

Mean = Sum of all values / Number of values


**Demonstrate the Concept**

**Numerical Example:**

* Marks of 5 students are 50, 70, 90, 40, 100.
* Sum = 50 + 70 + 90 + 40 + 100 = 350
* Number of students = 5
* Mean = 350 / 5 = 70

**Business Example:**

* A shop records daily sales of ₹10,000, ₹12,000, ₹8,000, ₹15,000, ₹5,000.
* The mean helps the shop owner understand the average daily sales.

**AI/ML Example:**

* In a customer dataset, mean can be used to find the average customer age.
* This helps us understand the data before building a Machine Learning model.


```python
# Import Librarics:
import numpy as np
import pandas as pd
```

```python
# Daily sales 
sales = np.array([10000, 12000, 8000, 15000, 5000])
# mean 
mean_sales = np.mean(sales)
print("Mean Daily Sales:", mean_sales)
```

## **2. Median**

**Understand the Concept**

* Median is the middle value in a group of numbers.
* First, arrange the values from smallest to largest.
* If there is an even number of values, take the average of the two middle values.

**Formula**

* For odd number of values:
    * Median = Middle value

* For even number of values:
    * Median = (Two middle values) / 2

**Demonstrate the Concept**

**Numerical Example:**

* Data: 10, 20, 30, 40, 50
* The middle value is 30.
* So, the median is 30.

**Business Example:**

* A company checks the salaries of its employees.
* Median salary helps the company understand the middle salary without being heavily affected by very high or very low salaries.

**AI/ML Example:**

* Median can be used to fill missing values in a dataset.
* It is useful when the data contains extreme values or outliers.

```python
# Implement the Concept:
# Salaries data
salaries = np.array([25000, 30000, 35000, 40000, 150000])
# Median 
median_salary = np.median(salaries)
print("Median Salary:", median_salary)
```

## **3. Mode**

**Understand the Concept**

* Mode is the value that appears most often in a group of data.
* A dataset can have one mode, more than one mode, or no mode.
* It is useful for finding the most common value.

**Formula**

Mode = Most frequently occurring value


**Demonstrate the Concept**

**Numerical Example:**

* Data: 10, 20, 20, 30, 40
* The value 20 appears two times.
* Other values appear only once.
* So, the mode is 20.

**Business Example:**

* A clothing shop checks the shoe sizes sold in a week.
* If size 8 is sold more than other sizes, size 8 is the mode.
* The shop can keep more size 8 in stock.

**AI/ML Example:**

* Mode can be used to fill missing values in categorical data.
* For example, if the most common city is Chennai, missing city values can be filled with Chennai.

```python
# Implement the Concept:
# Shoe sizes data
shoe_sizes = pd.Series([7, 8, 9, 8, 10, 8, 7, 9])
mode_size = shoe_sizes.mode()[0]  # [0] - show 1st row
print("Most Sold Shoe Size:", mode_size)
```

## **4. Weighted Mean**

**Understand the Concept**

* Weighted mean is an average where some values have more importance than others.
* Each value is multiplied by its weight.
* Then, the total is divided by the sum of the weights.

**Formula**

Weighted Mean = Σ(Value × Weight) / ΣWeight


**Demonstrate the Concept**

**Numerical Example:**

* Marks: 80, 70, 90
* Weights: 2, 3, 5

Weighted Mean:

* (80 × 2) + (70 × 3) + (90 × 5) = 830
* Total weight = 2 + 3 + 5 = 10
* Weighted Mean = 830 / 10 = 83

**Business Example:**

* A company gives different importance to interview, technical test, and communication.
* The technical test may have a higher weight than communication.
  
**AI/ML Example:**

* Weighted mean can be useful when some data points are more important than others.
* It can also be used when calculating weighted features or evaluation scores.


```python
# Implement the Concept:
# Marks & weights
marks = np.array([80, 70, 90])
weights = np.array([2, 3, 5])

# Weighted mean
weighted_mean = np.average(marks, weights=weights)

print("Weighted Mean:", weighted_mean)
```

## **5. Geometric Mean**

**Understand the Concept**

* Geometric mean is an average used for values that are multiplied together.
* It is useful for finding average growth rates and percentage changes.

**Formula**

Geometric Mean = ⁿ√(x₁ × x₂ × x₃ ... × xₙ)


**Demonstrate the Concept**

**Numerical Example:**

* Data: 2, 8
* Geometric Mean = √(2 × 8)
* Geometric Mean = √16
* Geometric Mean = 4

**Business Example:**

* A company has sales growth of 10% in one year and 20% in the next year.
* Geometric mean can be used to find the average growth rate over these years.

**AI/ML Example:**

* Geometric mean can be useful when combining growth rates or ratios.
* It can also be used in some ML evaluation methods where values are multiplied or have different scales.

```python
# Implement the Concept:
# Growth factors for two years
growth = np.array([1.10, 1.20])

# Geometric mean
geometric_mean = np.prod(growth) ** (1 / len(growth))

print("Geometric Mean:", geometric_mean)
print("Average Growth Rate:", (geometric_mean - 1) * 100, "%")
```

## **6. Harmonic Mean**

**Understand the Concept**

* Harmonic mean is an average used mainly for rates and ratios.
* It is useful when the same distance, work, or quantity is involved with different speeds or rates.
* It gives more importance to smaller values compared to a normal mean.

**Formula**

Harmonic Mean = n / (1/x₁ + 1/x₂ + ... + 1/xₙ)


**Demonstrate the Concept**

**Numerical Example:**

* A car travels the same distance at 60 km/h and 40 km/h.
* Harmonic Mean = 2 / (1/60 + 1/40)
* Harmonic Mean = 48 km/h

**Business Example:**

* A delivery company checks the delivery speed of two drivers for the same distance.
* Harmonic mean can be used to find the average speed.

**AI/ML Example:**

* Harmonic mean is used in Machine Learning for the F1-score.
* F1-score combines precision and recall into one value.
* It gives more importance when either precision or recall is low.

```python
# Implement the Concept:
from scipy.stats import hmean
# Speeds
speeds = [60, 40]
# Harmonic mean
harmonic_mean = hmean(speeds)
print("Harmonic Mean:", harmonic_mean)
```
