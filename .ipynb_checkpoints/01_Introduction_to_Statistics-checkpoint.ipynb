# **Notebook 1 – Introduction to Statistics**

## **1. What is Statistics?**

**Understand the Concept**
* It's a abstract science.
* It's the process of collecting, organizing, analyzing, and interpreting data.
* It helps us understand data, find patterns, and make better decisions.


**Demonstrate the Concept**

**Real-Life Example:-**

* A shop owner records daily sales.
* By using statistics, the owner can find the average daily sales and identify which days have higher or lower sales.


## **2.Why Statistics is important in AI/ML**

**Understand the Concept**

* Statistics helps us understand data before building a Machine Learning model.
* It helps us find patterns, check unusual values, and understand the data better.
* It also helps us check how well a model is working.

**Demonstrate the Concept**

**AI/ML Example:**

* Suppose we have customer data with age, salary, and spending.
* Using statistics, we can find the average age, average salary, and spending pattern of the customers.


## **3. Statistics vs Machine Learning**



**Understand the Concept**

**Statistics**
 * Statistics focuses more on understanding the data.
 * It mainly helps us understand and analyze data.

   
**Machine Learning**
 * Machine Learning focuses more on making predictions or decisions.
 * Its uses data to find patterns and make predictions.




**Demonstrate the Concept**

**Simple Example:**

* A shop owner uses statistics to find the average daily sales.
* A Machine Learning model can use past sales data to predict future sales.

**AI/ML Example:**

* Statistics helps us understand customer data.
* Machine Learning uses the same data to predict which customers may buy a product.

## **4. Descriptive Statistics**

**Understand the Concept**

* Descriptive statistics helps us summarize data in a simple way.
* It helps us understand the main information in a dataset.
* We can use numbers and graphs to describe the data.
* Mean, median, mode, minimum, maximum, and standard deviation are common measures.



**Demonstrate the Concept**

**Real-Life Example:**

* A gym records the number of steps taken by its members in a day.
* By using descriptive statistics, the gym can find the average steps, lowest steps, and highest steps.
* This helps understand the daily activity of the members.


```python
# Implement the Concept:
import pandas as pd

steps = pd.Series([5000, 7000, 8500, 6000, 10000])
print(steps.describe())
```

## **5. Inferential Statistics**

**Understand the Concept**

* Inferential statistics helps us understand a large group by studying a smaller group.
* We take a sample and use its results to make an idea about the whole population.
* It is useful when checking everyone takes too much time or effort.




**Demonstrate the Concept**

**Real-Life Example:**

* A college wants to know how many students are satisfied with the canteen food.
* Instead of asking every student, the college asks 100 students.
* The answers from these students can give an idea about the satisfaction of all students.

```python
# Implement the Concept:
import numpy as np
# sample data
marks = np.array([65, 70, 75, 80, 85, 90, 72, 68, 78, 82])
sample_mean = np.mean(marks)

print("Sample Mean:", sample_mean)
```

## **6. Population**

**Understand the Concept**

* Population means the complete group we want to study.
* It can include people, objects, or observations.
* The population contains all the data related to the study.

**Demonstrate the Concept**

**Real-Life Example:**

* A school wants to study the marks of all students in the school.
* If the school has 5 students in a small example, all 5 students are the population.
* We can study their marks to understand the performance of the whole group.



```python
# Implement the Concept:
import numpy as np

# Marks of all students
student_marks = np.array([65, 72, 80, 75, 90])
population_mean = np.mean(student_marks)

print("Student Marks:", student_marks)
print("Population Size:", len(student_marks))
print("Population Mean:", population_mean)
```

## **7. Sample**

**Understand the Concept**

* A sample is a small part taken from a population.
* We use a sample when studying the complete population is difficult.

**Demonstrate the Concept**

**Real-Life Example:**

* A school has 1,000 students and wants to know their average height.
* Instead of checking all 1,000 students, the school selects 100 students.
* These 100 students are the sample.

```python
# Implement the Concept:
# Sample - Heights 
sample_heights = np.array([160, 165, 170, 172, 168, 175, 162, 169, 171, 166])
sample_mean = np.mean(sample_heights)

print("Sample Heights:", sample_heights)
print("Sample Size:", len(sample_heights))
print("Sample Mean:", sample_mean)
```

## **8. Census**


**Understand the Concept**

* Census means collecting information from every member of a population.
* It gives complete information about the whole population.
* It can take more time and effort when the population is very large.

**Demonstrate the Concept**

**Real-Life Example:**

* A school wants to know the age of every student.
* Instead of selecting some students, the school collects the age of all students.
* This is called a census.


## **9. Sampling**

**Understand the Concept**

* Sampling means selecting a small group from a larger population.
* We use sampling when it is difficult or costly to study everyone.
  
**Demonstrate the Concept**

**Real-Life Example:**

* A company has 5,000 customers and wants to know their satisfaction level.
* Instead of asking all 5,000 customers, it selects 200 customers.
* These 200 customers are used as the sample.


```python
# Implement the Concept:
# Sample - customer satisfaction scores 
satisfaction_scores = np.array([4, 5, 3, 4, 5, 2, 4, 5, 3, 4])
average_score = np.mean(satisfaction_scores)

print("Sample Scores:", satisfaction_scores)
print("Sample Size:", len(satisfaction_scores))
print("Average Satisfaction Score:", average_score)
```

## **10. Variables**

**Understand the Concept**

* A variable is any characteristic that can have different values.
* It is used to store information about people, objects, or events.
* Examples include age, height, salary, and marks.

**Demonstrate the Concept**

**Real-Life Example:**

* A school records information about its students.
* Student age, height, and marks can be different for each student.
* These changing values are called variables.


```python
# Implement the Concept:
import pandas as pd
# Student info
students = pd.DataFrame({"Age": [18, 19, 20, 18, 21],
                         "Height": [165, 170, 175, 160, 172],
                         "Marks": [75, 82, 90, 68, 85]})
print(students)
```

## **11. Features**

**Understand the Concept**

* Features are input values given to a Machine Learning model.
* The model uses these values to make predictions.
* Features are also called independent variables.
* Examples include age, salary, experience, and house area.

**Demonstrate the Concept**

**Real-Life Example:**

* A company wants to predict the salary of an employee.
* The employee's experience, education, and age can be used as features.
* The ML model uses these features to predict the salary.


```python
# Implement the Concept:
# Employee data
employees = pd.DataFrame({"Experience": [1, 3, 5, 7, 10],
                          "Education_Years": [16, 16, 18, 18, 20],
                          "Age": [22, 25, 28, 32, 35],
                          "Salary": [25000, 35000, 45000, 60000, 80000]})


X = employees[["Experience", "Education_Years", "Age"]] # Features
y = employees["Salary"] # Target
print(X)
```

## **12. Observation**

**Understand the Concept**

* An observation is one individual record in a dataset.
* It contains information about one person, object, or event.
* In a table, one row usually represents one observation.

**Demonstrate the Concept**

**Real-Life Example:**

* A shop records customer details like age, order value, and rating.
* The details of one customer make one observation.
* Each customer will have a separate observation.

```python
# Implement the Concept:
# Customer data
customers = pd.DataFrame({"Age": [22, 28, 35, 31],
                          "Order_Value": [500, 750, 1200, 900],
                          "Rating": [4, 5, 3, 4]})
customers.iloc[0] # singlr row
```

## **13. Dataset**

**Understand the Concept**

* A dataset is a collection of related data.
* It contains rows and columns.
* Each row represents an observation and each column represents a variable or feature.

**Demonstrate the Concept**

**Real-Life Example:**

* A school stores student details such as name, age, marks, and attendance.
* All these details together form a dataset.
* The dataset helps the school understand student performance.

```python
# Implement the Concept:
# Student dataset
students = pd.DataFrame({"Name": ["Arun", "Bala", "Kumar", "Ravi"],
                         "Age": [18, 19, 18, 20],
                         "Marks": [75, 85, 68, 90],
                         "Attendance": [85, 92, 78, 95]})

print(students)
```

## **14. Quantitative Data**

**Understand the Concept**

* Quantitative data is data that can be measured or counted using numbers.
* We can perform mathematical calculations on this type of data.
* Examples include age, height, weight, salary, and marks.

**Demonstrate the Concept**

**Real-Life Example:**

* A gym records the weight of its members.
* The weights are given as numbers like 60 kg, 72 kg, and 85 kg.
* These values are quantitative data.


```python
# Implement the Concept:
# Weight 
weights = pd.Series([60, 72, 85, 68, 75])

print("Weights:", weights.tolist())
print("Average Weight:", weights.mean())
```

## **15. Qualitative Data**

**Understand the Concept**

* Qualitative data describes qualities or categories.
* It is usually not represented by numbers.
* Examples include color, gender, city, product type, and customer feedback.

**Demonstrate the Concept**

**Real-Life Example:**

* A restaurant records the type of food ordered by customers.
* The data can be Pizza, Burger, Biryani, or Dosa.
* These are categories, so they are qualitative data.

```python
# Implement the Concept:
# Food choices 
food = pd.Series(["Pizza", "Burger", "Dosa", "Pizza", "Biryani"])
print(food)

print("Most Ordered Food:", food.mode()[0])
```

## **16. Continuous Data**

**Understand the Concept**

* Continuous data can have any value within a range.
* It is usually measured, not counted.
* Examples include height, weight, temperature, and distance.

**Demonstrate the Concept**

**Real-Life Example:**

* A doctor checks the weight of patients.
* A person's weight can be 65 kg, 65.5 kg, or 65.75 kg.
* So, weight is continuous data.

```python
# Implement the Concept:
# Weight
weights = pd.Series([65.2, 70.5, 68.7, 72.3, 60.8])
print(weights)

print("Average Weight:", weights.mean())
```

## **17. Discrete Data**

**Understand the Concept**

* Discrete data is data that can be counted.
* It usually contains whole numbers.
* It cannot normally have decimal values.
* Examples include number of students, number of cars, and number of orders.

**Demonstrate the Concept**

**Real-Life Example:**

* A shop counts the number of customers visiting each day.
* The number can be 50, 75, or 100 customers.
* We cannot have 75.5 customers, so the number of customers is discrete data.

```python
# Implement the Concept:
# No of customers visiting a shop
customers = pd.Series([50, 75, 60, 90, 80])
print(customers)
print("Total Customers:", customers.sum())
```

## **18. Structured Data**

**Understand the Concept**

* Structured data is data arranged in a fixed format.
* It is usually stored in rows and columns.

**Demonstrate the Concept**

**Real-Life Example:**

* A school stores student details in a table.
* The table has columns like Name, Age, Marks, and Attendance.
* This is structured data because the information follows a fixed format.


```python
# Implement the Concept:
# Student data
students = pd.DataFrame({"Name": ["Arun", "Bala", "Kumar", "Ravi"],
                         "Age": [18, 19, 20, 18],
                         "Marks": [75, 85, 90, 68]})
print(students)
```

## **19. Unstructured Data**



**Understand the Concept**

* Unstructured data does not have a fixed format like rows and columns.
* It can contain text, images, videos, audio, and documents.
* It is usually harder to organize and analyze directly.

**Demonstrate the Concept**

**Real-Life Example:**

* A company receives customer feedback through messages, photos, and voice recordings.
* These files do not follow a fixed table format.


```python
# Implement the Concept:
# unstructured data
customer_feedback = [ "The product is very good",
                      "Delivery was late",
                      "I liked the product"]
for feedback in customer_feedback:
    print(feedback)
```

## Interview Questions & Answers

**Q1. What is Statistics?**

**Answer:**

Statistics is the process of collecting, organizing, analyzing, and understanding data.

**Q2. Why is Statistics important in AI/ML?**

**Answer:**

Statistics helps us understand data, find patterns, detect unusual values, and make better decisions before building an ML model.

**Q3. What is the difference between Population and Sample?**

**Answer:**

Population is the complete group we want to study. Sample is a smaller part taken from that population.

**Q4. What is the difference between Quantitative and Qualitative data?**

**Answer:**

Quantitative data contains numbers, like age, salary, and marks. Qualitative data contains categories, like color, city, and product type.

**Q5. What is the difference between Continuous and Discrete data?**

**Answer:**

Continuous data can have decimal values, like weight and height. Discrete data is counted in whole numbers, like number of students or orders.

**Q6. What is the difference between Structured and Unstructured data?**

**Answer:**

Structured data is arranged in rows and columns. Unstructured data does not have a fixed format, like images, videos, and text.
