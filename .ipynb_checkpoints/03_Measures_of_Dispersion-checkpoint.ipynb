# **Notebook 3 – Measures of Dispersion**

**Measures of Dispersion**

Its a number that describes how spread out or scattered the values in a dataset are around a central value like the mean

**Why is it Important?**
* Checking Average
* Finding Consistency
* Spotting Outliers
* Comparing Groups
* Risk and Quality Control

## **1. Range**

**Understand the Concept**

* Range shows the difference between the highest and lowest value in a dataset.
* It gives a simple idea about how spread out the data is.

**Formula**

Range = Maximum Value - Minimum Value



**Demonstrate the Concept**

**Numerical Example:**

* Data: 10, 20, 30, 40, 50
* Maximum = 50
* Minimum = 10
* Range = 50 - 10 = 40

**Real-Life Example:**

* A teacher checks the marks of students.
* The highest mark is 95 and the lowest mark is 55.
* The range is 40 marks.

**AI/ML Example:**

* Range can help us understand how much a feature varies.
* For example, we can check the range of age in a customer dataset.


```python
# Implement the concept:
import numpy as np
# Marks
marks = np.array([10,20,30,40,50])
# Range 
mark_range = np.max(marks) - np.min(marks)
print("Highest Mark:", np.max(marks))
print("Lowest Mark:", np.min(marks))
print("Range:", mark_range)
```

## **2. Variance**

**Understand the Concept**

* Variance measures the average squared deviation of each data point from mean.
* A small variance means the values are close to the mean.
* A large variance means the values are more spread out.

**Formula**

Population Variance = Σ(x - μ)² / N

Where:
* x = each value
* μ = mean
* N = number of values


**Demonstrate the Concept**

**Numerical Example:**

* Data: 2, 4, 6
* Mean = (2 + 4 + 6) / 3 = 4
* Differences from mean = -2, 0, 2
* Squared differences = 4, 0, 4
* Variance = (4 + 0 + 4) / 3 = 2.67

**Real-Life Example:**

* A teacher compares the marks of two classes.
* If the marks in one class are close to the average, the variance is low.
* If the marks are very different from the average, the variance is high.

**AI/ML Example:**

* Variance helps us understand how much a feature changes in a dataset.
* It can also help identify features with very little variation.


```python
# Implement the Concept:
marks = np.array([2, 4, 6])
print("Mean:", np.mean(marks)) # Mean
print("Variance:", np.var(marks)) # Variance
```

## **3. Standard Deviation**

**Understand the Concept**

* Standard deviation is the square root of variance.
* A small standard deviation means the values are close to the mean.
* A large standard deviation means the values are more spread out.


**Formula**

Standard Deviation = √Variance


**Demonstrate the Concept**

**Numerical Example:**

* Data: 2, 4, 6
* Mean = 4
* Variance = 2.67
* Standard Deviation = √2.67 ≈ 1.63

**AI/ML Example:**

* Standard deviation helps us understand the spread of features.
* It is also useful when scaling data before training a Machine Learning model.

```python
print("Mean:", np.mean(marks)) # Mean
print("Variance:", np.var(marks)) # Variance
print("Standard Deviation:", np.std(marks)) # Standard deviation
```

## **4. Quartiles**

**Understand the Concept**

* Quartiles divide sorted data into four equal parts.
* There are three main quartiles: Q1, Q2, and Q3.
    * Q1 is the 25th percentile
    * Q2 is the median
    * Q3 is the 75th percentile.

**Formula**

* Q1 = 25th percentile
* Q2 = 50th percentile
* Q3 = 75th percentile



**Demonstrate the Concept**

**Numerical Example:**

* Data: 10, 20, 30, 40, 50, 60, 70, 80
* Q1 shows the value below which about 25% of the data falls.
* Q2 is the middle value of the data.
* Q3 shows the value below which about 75% of the data falls.

**Real-Life Example:**

* A company checks the salaries of its employees.
* Quartiles can divide the employees into four salary groups.
* This helps the company understand how salaries are distributed.

```python
# Implement the Concept:
# Employee salaries
salaries = np.array([20000, 25000, 30000, 35000, 40000, 45000, 50000, 60000])

# Quartiles
q1 = np.percentile(salaries, 25)
q2 = np.percentile(salaries, 50)
q3 = np.percentile(salaries, 75)
q4 = np.percentile(salaries, 100)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
print("Q4:", q4)
```

## **5. Percentiles**



**Understand the Concept**

* Percentile tells us the position of a value compared with other values.
* It shows the value below which a certain percentage of data falls.
* For example, the 75th percentile means about 75% of the data is below that value.




**Demonstrate the Concept**

**Numerical Example:**

* Data: 10, 20, 30, 40, 50, 60, 70, 80
* The 25th percentile is 27.5.
* This means about 25% of the data is below 27.5.

**Real-Life Example:**

* A student scores 85 marks in an exam.
* If the student is in the 90th percentile, it means the student performed better than about 90% of the students.

```python
# Implement the Concept:
# Exam marks
marks = np.array([10, 20, 30, 40, 50, 60, 70, 80])

# Percentiles
p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p75 = np.percentile(marks, 75)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
```

## **6. Interquartile Range (IQR)**



**Understand the Concept**

* IQR shows the spread of the middle 50% of the data.
* It is calculated using the third quartile and first quartile.

**Formula**

IQR = Q3 - Q1



**Demonstrate the Concept**

**Numerical Example:**

* Data: 10, 20, 30, 40, 50, 60, 70, 80
* Q1 = 27.5
* Q3 = 62.5
* IQR = 62.5 - 27.5 = 35

**Real-Life Example:**

* A company checks employee salaries.
* IQR helps the company understand the spread of the middle 50% of salaries.
* Very high or very low salaries can be checked as possible outliers.

**AI/ML Example:**

* IQR is commonly used to detect outliers in Machine Learning data.
* It can help find unusual values before training a model.


```python
# Implement the Concept:
# Salaries
salaries = np.array([20000, 25000, 30000, 35000, 40000, 45000, 50000, 60000])

# Q1 & Q3
q1 = np.percentile(salaries, 25)
q3 = np.percentile(salaries, 75)

# IQR
iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

image.png

```python
# Detecting outliers
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

# find and store outliers in a list using list comprehension
outliers = [x for x in salaries if x < lower_bound or x > upper_bound]
print("Outliers:", outliers)
```

## **7. Coefficient of Variation**

**Understand the Concept**

* Coefficient of Variation (CV) shows the amount of variation compared with the mean.
* It is useful when we want to compare the spread of two different datasets.
* A lower CV means the data is more consistent.

**Formula**

CV = (Standard Deviation / Mean) × 100


**Demonstrate the Concept**

**Numerical Example:**

* Mean = 50
* Standard Deviation = 5
* CV = (5 / 50) × 100
* CV = 10%

**Real-Life Example:**

* Two machines produce the same type of product.
* We can use CV to compare how consistent their production is.
* The machine with the lower CV has more consistent output.


```python
# Implement the Concept:
# Two machines
machine_a = np.array([48, 50, 52, 49, 51])
machine_b = np.array([40, 50, 60, 45, 55])

# Mean
mean_a = np.mean(machine_a)
mean_b = np.mean(machine_b)

# Std
std_a = np.std(machine_a)
std_b = np.std(machine_b)

# CV
cv_a = (std_a / mean_a) * 100
cv_b = (std_b / mean_b) * 100

print("Machine A CV:", cv_a, "%")
print("Machine B CV:", cv_b, "%")
```

Machine A has lower variation, so its production is more consistent.
