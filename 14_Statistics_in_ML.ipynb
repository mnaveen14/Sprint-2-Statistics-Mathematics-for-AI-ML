## **1. Data Cleaning**



**Understand the Concept**

* Data cleaning means finding and fixing problems in the data.
* We may have wrong values, duplicate records, or missing data.
* Statistics helps us understand the data and find unusual values.



**Demonstrate the Concept**

**Example:**

* A teacher records student ages as 18, 19, 20, 200, 21.
* The value 200 is clearly unusual.
* Statistics can help us find such values and check the data.

**AI/ML Example**

* Before training an ML model, data should be clean.
* We can use mean, median, standard deviation, and other statistical methods to understand and clean the data.


```python
# Libraries
import numpy as np
import pandas as pd
```

```python
# Implement the Concept:
# age data
ages = pd.DataFrame([18, 19, 20, 200, 21])

print(ages.describe())
```

**Interpretation**

* We found it while checking the data.

## **2. Missing Value Handling**



**Understand the Concept**

* Missing values are empty or unavailable values in a dataset.
* We need to handle them before using the data for Machine Learning.
* Statistics can help us fill missing values using mean, median, or mode.



**Demonstrate the Concept**

**Example:**

* A student dataset has marks like 70, 80, 75, missing, and 90.
* We can use the average of the available marks to fill the missing value.


```python
# Implement the Concept:
# Student marks data
marks = pd.Series([70, 80, 75, np.nan, 90])

print("Before Data:")
print(marks)

# Fill missing value  - using mean
mean_marks = marks.mean()
marks_filled = marks.fillna(mean_marks)

print("\nMean Marks:", mean_marks)
print("\nData After :")
print(marks_filled)
```

**Interpretation**

* One mark is missing from the data.
* The missing value is replaced with the mean.

## **3. Outlier Detection**



**Understand the Concept**

* Outlier detection means finding values that are very different from the other values.
* Statistics helps us identify these unusual values.
* Common methods are IQR and Z-Score.



**Demonstrate the Concept**

**Example:**

* A student gets marks of 60, 65, 70, 72, and 150.
* Most marks are between 60 and 72.
* The value 150 is very different, so it can be an outlier.



```python
# Implement the Concept:
from scipy.stats import zscore

# marks
marks = np.array([60, 65, 70, 72, 150])

# Z-Scores
z_scores = zscore(marks)

# Find possible outliers
outliers = marks[np.abs(z_scores) > 1.5]

print("Marks:", marks)
print("Z-Scores:", z_scores)
print("Possible Outliers:", outliers)
```

**Interpretation**

* Most marks are between 60 and 72.
* The value 150 is much higher than the other marks.

## **4. Feature Engineering**



**Understand the Concept**

* Feature engineering means creating or changing features to make them more useful for an ML model.
* Statistics helps us understand the data before creating new features.
* We can use mean, median, ratios, and other statistical values.



**Demonstrate the Concept**

**Example:**

* A shop has the number of items bought and the total amount spent by each customer.
* We can create a new feature called Average Price Per Item.
* It is calculated using total amount divided by number of items.

**AI/ML Example**

* Feature engineering can help an ML model get better information from the existing data.
* Statistical calculations can be used to create useful new features.



```python
# Implement the Concept:
# Customer purchase data
data = pd.DataFrame({"Items_Bought": [2, 4, 5, 10],
                     "Total_Spending": [200, 360, 550, 900]})

# Create a new feature
data["Average_Price_Per_Item"] = (data["Total_Spending"] / data["Items_Bought"])

print(data)
```

## **5. Feature Scaling**



**Understand the Concept**

* Feature scaling means bringing different features to a similar scale.
* It is useful when features have very different values.
* It helps some ML models work better and faster.



**Demonstrate the Concept**

**Example:**

* Suppose we have a student's age and monthly study hours.
* Age may be between 18 and 25, while study hours may be between 100 and 500.
* Since the values have different ranges, scaling can bring them to a similar range.

**AI/ML Example**

* Feature scaling is important for algorithms like KNN, K-Means, SVM, and Logistic Regression.
* It prevents features with larger values from having too much influence on the model.



```python
# Implement the Concept:
from sklearn.preprocessing import StandardScaler

# Student data
data = pd.DataFrame({"Age": [18, 20, 22, 24, 25],
                     "Study_Hours": [100, 200, 300, 400, 500]})

# Standard Scaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

# Convert back to DataFrame
scaled_data = pd.DataFrame(scaled_data, columns=data.columns)

print("Original Data:")
print(data)

print("\nScaled Data:")
print(scaled_data)
```

## **6. Data Normalization**



**Understand the Concept**

* Data normalization means changing values to a common range.
* It is useful when features have very different values.
* A common range is 0 to 1.
* Min-Max scaling is a common method for normalization.



**Demonstrate the Concept**

**Example:**

* A student has marks between 0 and 100.
* Another feature like attendance is between 0 and 100 as well, but in other datasets features can have very different ranges.
* Normalization brings the values to a common range so they can be compared more easily.

**AI/ML Example**

* Normalization is useful for ML algorithms that depend on distance or feature scale.
* It is commonly used in KNN, K-Means, and neural networks.


```python
# Implement the Concept:
from sklearn.preprocessing import MinMaxScaler

# Student data
data = pd.DataFrame({"Marks": [40, 50, 60, 70, 80],
                     "Study_Hours": [2, 4, 6, 8, 10]})

# Min-Max scaler
scaler = MinMaxScaler()
normalized_data = scaler.fit_transform(data)

normalized_data = pd.DataFrame(normalized_data, columns=data.columns)

print("Original Data:")
print(data)

print("\nNormalized Data:")
print(normalized_data)
```

## **7. Model Evaluation**



**Understand the Concept**

* Model evaluation means checking how well a Machine Learning model is performing.
* Statistics helps us measure the model's performance using different metrics.
* The metric we use depends on the type of ML problem.



**Demonstrate the Concept**

**Example:**

* A model predicts whether students will pass or fail.
* We compare the model's predictions with the actual results.

**AI/ML Example**

* For classification, we can use Accuracy, Precision, Recall, and F1-Score.
* For regression, we can use MAE, MSE, and RMSE.




```python
# Implement the Concept:
from sklearn.metrics import accuracy_score

# Actual results
actual = [1, 1, 0, 1, 0, 0, 1, 1]

# Model predictions
predicted = [1, 1, 0, 0, 0, 0, 1, 1]

# accuracy
accuracy = accuracy_score(actual, predicted)

print("Model Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

## **8. Linear Regression**



**Understand the Concept**

* Linear Regression is used to predict a numerical value.
* It finds a relationship between input features and the target value.
* The model tries to fit a straight line through the data.



**Demonstrate the Concept**

**Example:**

* A person wants to predict house rent based on house area.
* Larger houses usually have higher rent.
* Linear Regression can learn this relationship and predict the rent.

**AI/ML Example**

* Linear Regression is used for predicting values like house prices, sales, salary, and temperature.
* Statistics helps us understand the relationship between the features and target value.



```python
# Implement the Concept:
from sklearn.linear_model import LinearRegression

# House area & monthly rent
area = np.array([500, 700, 900, 1100, 1300]).reshape(-1, 1)
rent = np.array([8000, 10000, 12000, 14000, 16000])

# train the model
model = LinearRegression()
model.fit(area, rent)

# Predict rent for a 1000 sq.ft house
predicted_rent = model.predict([[1000]])

print("Predicted Rent:", predicted_rent[0])
```

## **9. Logistic Regression**



**Understand the Concept**

* Logistic Regression is used to predict two possible outcomes.
* The output is usually 0 or 1.
* It gives the probability of each outcome and then makes a prediction.



**Demonstrate the Concept**

**Example:**

* A student wants to know whether they will pass or fail an exam.
* Study hours can be used to predict the result.
* Pass = 1
* Fail = 0

**AI/ML Example**

* Logistic Regression is commonly used for binary classification.
* Examples include Spam/Not Spam, Yes/No, and Disease/No Disease.



```python
# Implement the Concept:
from sklearn.linear_model import LogisticRegression

# data
study_hours = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
results = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# train model
model = LogisticRegression()
model.fit(study_hours, results)

# Predict result - 5h study
prediction = model.predict([[5]])
probability = model.predict_proba([[5]])

print("Prediction:", prediction[0])
print("Probability of Fail:", probability[0][0])
print("Probability of Pass:", probability[0][1])
```

## **10. Naive Bayes**



**Understand the Concept**

* Naive Bayes is a Machine Learning algorithm based on probability.
* It uses Bayes' Theorem to predict the class of new data.
* It assumes that the features are independent of each other.



**Demonstrate the Concept**

**Example:**

* A bank checks a customer's age, income, and loan history.
* Naive Bayes can use this information to predict whether the customer will repay the loan.


```python
# Implement the Concept:
from sklearn.naive_bayes import GaussianNB

# Customer data - age, income, loan history
X = [[22, 25, 0],
    [25, 30, 0],
    [35, 50, 1],
    [40, 60, 1],
    [45, 70, 1],
    [28, 35, 0]]

# 0 =  Not Repay, 1 = Repay
y = [0, 0, 1, 1, 1, 0]

# train model
model = GaussianNB()
model.fit(X, y)

# New customer
new_customer = [[38, 55, 1]]

# Predict loan repayment
prediction = model.predict(new_customer)

if prediction[0] == 1:
    print("Prediction: Will Repay")
else:
    print("Prediction: Will Not Repay")
```

## **11. Clustering**



**Understand the Concept**

* Clustering is used to group similar data points together.
* It does not need predefined labels.
* Data points in the same group are more similar to each other.



**Demonstrate the Concept**

**Example:**

* A shop has customer data based on age and spending.
* Clustering can group customers with similar buying behavior.
* This helps the shop understand different types of customers.


```python
# Implement the Concept:
from sklearn.cluster import KMeans

# Customer data - Age, Monthly Spending
customers = np.array([[20, 2000],
                      [22, 2500],
                      [25, 3000],
                      [40, 7000],
                      [42, 7500],
                      [45, 8000]])

# Create 2 groups
model = KMeans(n_clusters=2, random_state=42, n_init=10)

# Train the model
model.fit(customers)

# cluster labels
labels = model.labels_

print("Customer Groups:", labels)
```

## **12. Principal Component Analysis (PCA)**



**Understand the Concept**

* PCA is used to reduce the number of features in a dataset.
* It combines related features into new features called principal components.
* It helps keep the important information while reducing the data size.



**Demonstrate the Concept**

**Example:**

* A hospital has patient data like age, weight, blood pressure, and heart rate.
* Some of these features may be related.
* PCA can reduce these features into fewer components while keeping the main information.


```python
# Implement the Concept:
from sklearn.decomposition import PCA

# Patient data - Age, Weight, Blood Pressure, Heart Rate
data = np.array([[25, 60, 120, 72],
                 [30, 65, 125, 75],
                 [35, 70, 130, 78],
                 [40, 75, 135, 80],
                 [45, 80, 140, 82]])

# Scale the data
scaled_data = StandardScaler().fit_transform(data)

# Reduce 4 features to 2 components
pca = PCA(n_components=2)
reduced_data = pca.fit_transform(scaled_data)

print("Original Shape:", data.shape)
print("Reduced Shape:", reduced_data.shape)
print("Reduced Data:")
print(reduced_data)
```

## **13. Recommendation Systems**

**Understand the Concept**

* Recommendation systems suggest products, movies, songs, or other items based on user data.
* Statistics helps find patterns in user choices and preferences.
* These patterns are used to give better recommendations.





**Demonstrate the Concept**

**Example:**

* A movie app checks which movies a user has watched and rated.
* If a user likes action movies, the app can suggest similar movies.



```python
# Implement the Concept:
# movie ratings
data = pd.DataFrame({"Movie": ["Movie A", "Movie B", "Movie C", "Movie D"],
                     "Rating": [5, 4, 2, 5]})

#  avg rating
average_rating = data["Rating"].mean()

# Recommend movies with rating >= avg
recommendations = data[data["Rating"] >= average_rating]

print("Average Rating:", average_rating)
print("\nRecommended Movies:")
print(recommendations)
```

This is a example of how user data can be used to recommend items.
